```python
from transformers import VitsModel, AutoTokenizer
from IPython.display import Audio
import torch
import scipy
```

```python
model_rus = VitsModel.from_pretrained("facebook/mms-tts-rus")
model_translit = VitsModel.from_pretrained("facebook/mms-tts-rus")
model_eng = VitsModel.from_pretrained("facebook/mms-tts-rus")
model_eng_translit = VitsModel.from_pretrained("facebook/mms-tts-rus")

tokenizer_rus = AutoTokenizer.from_pretrained("facebook/mms-tts-rus")
tokenizer_translit = AutoTokenizer.from_pretrained("facebook/mms-tts-rus")
tokenizer_eng = AutoTokenizer.from_pretrained("facebook/mms-tts-rus")
tokenizer_eng_translit = AutoTokenizer.from_pretrained("facebook/mms-tts-rus")

text_rus = "Привет! Я учусь в Гиг Брайнс работать с искусственным интеллектом."
text_translit = "Privet! Ya uchus v GeekBrains rabotat s iskusstvennim intellektom."
text_eng = "Hello! I am studying at GeekBrains to work with artificial intelligence."
text_eng_translit = "Хеллоу! Ай эм стадинг эт Гиг Брайнс то ворк виз артификал интеленджез."

inputs_rus = tokenizer_rus(text_rus, return_tensors="pt")
inputs_translit = tokenizer_translit(text_translit, return_tensors="pt")
inputs_eng = tokenizer_eng(text_eng, return_tensors="pt")
inputs_eng_translit = tokenizer_eng_translit(text_eng_translit, return_tensors="pt")

with torch.no_grad():
    output_rus = model_rus(**inputs_rus).waveform
    output_translit = model_translit(**inputs_translit).waveform
    output_eng = model_eng(**inputs_eng).waveform
    output_eng_translit = model_eng(**inputs_eng_translit).waveform
```

```python
### русская речь из текста на русском языке
Audio(output_rus, rate=model_rus.config.sampling_rate)
```

```python
### русская речь из русского текста на транслите
Audio(output_translit, rate=model_translit.config.sampling_rate)
```

```python
### английская речь из текста на английском языке
Audio(output_eng, rate=model_eng.config.sampling_rate)
```

```python
### английская речь из текста на русской транслитерации
Audio(output_eng_translit, rate=model_eng_translit.config.sampling_rate)
```
