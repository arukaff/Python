```python
import numpy as np
import tensorflow as tf 
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, SimpleRNN
```

```python
text='This is GeekBrains neural net arhitecture course!'
chars = sorted(list(set(text)))
char_to_index = {char: i for i, char in enumerate(chars) }
index_to_char = {i: char for i, char in enumerate(chars)}
```

```python
len(chars)
```

```python
seq_leght =4
sequence=[]
labels =[]
for i in range(0,len(text)-seq_leght):
    seq=text[i:i+seq_leght]
    label= text[i+seq_leght]
    sequence.append([char_to_index[char] for char in seq])
    labels.append(char_to_index[label])
```

```python
sequence
```

```python
labels
```

```python
X=np.array(sequence)
y=np.array(labels)
```

```python
X
```

```python
y
```

```python
X_one_hot=tf.one_hot(X,len(chars))
y_one_hot=tf.one_hot(y,len(chars))
```

```python
X_one_hot
```

```python
y_one_hot
```

```python
model=Sequential()
model.add(SimpleRNN(50,input_shape=(seq_leght,len(chars)),activation='relu'))
model.add(Dense(len(chars),activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy', optimizer='adam',metrics=['accuracy'])
```

```python
model.fit(X_one_hot,y_one_hot,epochs=100)
```

```python
start_seq='This is G'
gen_text=start_seq
for i in range(50):
    x=np.array([[char_to_index[char] for char in gen_text[-seq_leght:]]])
    print(x)
    x_one_hot= tf.one_hot(x,len(chars))
    predict=model.predict(x_one_hot)
    print(predict)
    next_index=np.argmax(predict)
    print(next_index)
    next_char=index_to_char[next_index]
    print(next_char)
    gen_text+=next_char
    print(gen_text)
print(gen_text)
```
