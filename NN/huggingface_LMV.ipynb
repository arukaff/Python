```python
from transformers import ViTFeatureExtractor, ViTForImageClassification
from PIL import Image
import requests
url = 'https://cdn.f1ne.ws/userfiles/143793.jpg'
image = Image.open(requests.get(url, stream=True).raw)
feature_extractor = ViTFeatureExtractor.from_pretrained('google/vit-base-patch16-384')
model = ViTForImageClassification.from_pretrained('google/vit-base-patch16-384')
inputs = feature_extractor(images=image, return_tensors="pt")
outputs = model(**inputs)
logits = outputs.logits
# model predicts one of the 1000 ImageNet classes
predicted_class_idx = logits.argmax(-1).item()
print("Predicted class:", model.config.id2label[predicted_class_idx])
```

```python
from transformers import DetrImageProcessor, DetrForObjectDetection
import torch
from PIL import Image
import requests

url = "http://images.cocodataset.org/val2017/000000039769.jpg"
#url = 'https://cdn.f1ne.ws/userfiles/143793.jpg'
image = Image.open(requests.get(url, stream=True).raw)

# you can specify the revision tag if you don't want the timm dependency
processor = DetrImageProcessor.from_pretrained("facebook/detr-resnet-50", revision="no_timm")
model = DetrForObjectDetection.from_pretrained("facebook/detr-resnet-50", revision="no_timm")

inputs = processor(images=image, return_tensors="pt")
outputs = model(**inputs)

# convert outputs (bounding boxes and class logits) to COCO API
# let's only keep detections with score > 0.9
target_sizes = torch.tensor([image.size[::-1]])
results = processor.post_process_object_detection(outputs, target_sizes=target_sizes, threshold=0.5)[0]

for score, label, box in zip(results["scores"], results["labels"], results["boxes"]):
    box = [round(i, 2) for i in box.tolist()]
    print(
            f"Detected {model.config.id2label[label.item()]} with confidence "
            f"{round(score.item(), 3)} at location {box}"
    )
```

```python
from transformers import VitsModel, AutoTokenizer
from IPython.display import Audio
import torch
import scipy
```

```python
model_rus = VitsModel.from_pretrained("facebook/mms-tts-rus")
model_translit = VitsModel.from_pretrained("facebook/mms-tts-rus")
model_eng = VitsModel.from_pretrained("facebook/mms-tts-rus")
model_eng_translit = VitsModel.from_pretrained("facebook/mms-tts-rus")

tokenizer_rus = AutoTokenizer.from_pretrained("facebook/mms-tts-rus")
tokenizer_translit = AutoTokenizer.from_pretrained("facebook/mms-tts-rus")
tokenizer_eng = AutoTokenizer.from_pretrained("facebook/mms-tts-rus")
tokenizer_eng_translit = AutoTokenizer.from_pretrained("facebook/mms-tts-rus")

text_rus = "Привет! Я учусь в Гиг Брайнс работать с искусственным интеллектом."
text_translit = "Privet! Ya uchus v GeekBrains rabotat s iskusstvennim intellektom."
text_eng = "Hello! I am studying at GeekBrains to work with artificial intelligence."
text_eng_translit = "Хеллоу! Ай эм стадинг эт Гиг Брайнс то ворк виз артификал интеленджез."

inputs_rus = tokenizer_rus(text_rus, return_tensors="pt")
inputs_translit = tokenizer_translit(text_translit, return_tensors="pt")
inputs_eng = tokenizer_eng(text_eng, return_tensors="pt")
inputs_eng_translit = tokenizer_eng_translit(text_eng_translit, return_tensors="pt")

with torch.no_grad():
    output_rus = model_rus(**inputs_rus).waveform
    output_translit = model_translit(**inputs_translit).waveform
    output_eng = model_eng(**inputs_eng).waveform
    output_eng_translit = model_eng(**inputs_eng_translit).waveform
```

```python
### русская речь из текста на русском языке
Audio(output_rus, rate=model_rus.config.sampling_rate)
```

```python
### английская речь из текста на русской транслитерации
Audio(output_eng_translit, rate=model_eng_translit.config.sampling_rate)
```
